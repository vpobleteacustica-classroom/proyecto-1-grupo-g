### Lógica de Generación MIDI
La siguiente función `generar_midi_desde_pgn` toma el texto PGN y lo convierte en música.
* **Mapeo:** Cada pieza tiene una nota base (Peón = Do, Rey = La, etc.).
* **Tablero:** Las columnas (files) controlan el paneo (izquierda/derecha) y las filas (ranks) controlan la octava.
* **Eventos:** Capturas y Jaques tienen mayor "Velocity" (volumen).

```python
# ----------------------------------------------------------------------
#  Función: generar el MIDI desde el PGN con parámetros
# ----------------------------------------------------------------------
def generar_midi_desde_pgn(pgn_texto, tempo_bpm, prog_blancas, prog_negras,
                           output_filename='partida_sonificada_blancas_negras.mid'):

    pgn_io = io.StringIO(pgn_texto)
    game = chess.pgn.read_game(pgn_io)

    if not game:
        raise ValueError("No se pudo cargar la partida PGN.")

    print("¡Partida PGN cargada exitosamente!")
    print(f"Partida: {game.headers.get('Event', 'Sin nombre')}")

    # 1. Mapeo de Piezas a Nota MIDI 
    mapa_piezas_nota = {
        chess.PAWN:   60,  # Do4
        chess.KNIGHT: 62,  # Re4
        chess.BISHOP: 64,  # Mi4
        chess.ROOK:   65,  # Fa4
        chess.QUEEN:  67,  # Sol4
        chess.KING:   69,  # La4
    }

    # 2. Parámetros de síntesis
    DURACION_NOTA   = 240
    VELOCITY_NORMAL = 80
    VELOCITY_CAPTURA = 110
    VELOCITY_JAQUE   = 127
    DURACION_ARPEGIO = 120
    VELOCITY_ESPECIAL = 127

    print("Mapeo de sonidos definido.")

    # 3. Mapeo de Columna (a-h) a Paneo MIDI (0-127)
    mapa_columna_pan = {
        0: 0,
        1: 18,
        2: 36,
        3: 54,
        4: 72,
        5: 90,
        6: 108,
        7: 127
    }

    # 4. Mapeo de Fila (1-8) a offset de nota
    mapa_fila_octava = {
        0: 0,
        1: 2,
        2: 4,
        3: 6,
        4: 8,
        5: 10,
        6: 12,
        7: 14
    }

    print("Mapeo de Filas y Columnas definido.")

    # 1. Configurar el archivo MIDI
    mid = mido.MidiFile()  # ticks_per_beat por defecto = 480

    track_blancas = mido.MidiTrack()
    track_negras  = mido.MidiTrack()
    mid.tracks.append(track_blancas)
    mid.tracks.append(track_negras)

    # Metadatos
    track_blancas.append(mido.MetaMessage('track_name', name='Sonificación Blancas ACUS220'))
    track_negras.append(mido.MetaMessage('track_name', name='Sonificación Negras ACUS220'))

    # Tempo (parametrizable)
    tempo_val = mido.bpm2tempo(tempo_bpm)
    track_blancas.append(mido.MetaMessage('set_tempo', tempo=tempo_val))
    track_negras.append(mido.MetaMessage('set_tempo', tempo=tempo_val))

    # Instrumentos (parametrizables)
    track_blancas.append(mido.Message('program_change', program=prog_blancas, channel=0, time=0))
    track_negras.append(mido.Message('program_change', program=prog_negras,  channel=1, time=0))

    # 3. Iterar sobre la partida y generar eventos
    board = game.board()
    print("Iniciando sonificación...")

    for move in game.mainline_moves():

        pieza = board.piece_at(move.from_square)
        if pieza is None:
            board.push(move)
            continue

        # Elegir pista y canal según el color
        if pieza.color == chess.WHITE:
            track_actual = track_blancas
            canal = 0
        else:
            track_actual = track_negras
            canal = 1

        # Coordenadas del movimiento
        columna_destino = chess.square_file(move.to_square)
        fila_destino    = chess.square_rank(move.to_square)

        pan_valor     = mapa_columna_pan.get(columna_destino, 64)
        octava_offset = mapa_fila_octava.get(fila_destino, 0)

        # paneo
        track_actual.append(
            mido.Message('control_change', channel=canal, control=10, value=pan_valor, time=0)
        )

        # --- Promoción ---
        if move.promotion:
            print(f"Promoción detectada: {move}")
            pieza_promovida = move.promotion
            nota_base = mapa_piezas_nota.get(pieza_promovida, 67)
            nota_final = nota_base + octava_offset

            # Acorde mayor
            for n in (nota_base, nota_base + 4, nota_base + 7):
                track_actual.append(
                    mido.Message('note_on', note=n,
                                 velocity=VELOCITY_ESPECIAL, time=0, channel=canal)
                )
            for n in (nota_base, nota_base + 4, nota_base + 7):
                track_actual.append(
                    mido.Message(
                        'note_off',
                        note=n,
                        velocity=0,
                        time=DURACION_NOTA * 2 if n == nota_base else 0,
                        channel=canal
                    )
                )

            board.push(move)
            continue

        # --- Enroque ---
        if board.is_castling(move):
            print(f"Enroque detectado: {move}")
            nota_rey   = mapa_piezas_nota.get(chess.KING) + octava_offset
            nota_torre = mapa_piezas_nota.get(chess.ROOK) + octava_offset

            track_actual.append(
                mido.Message('note_on', note=nota_rey,
                             velocity=VELOCITY_ESPECIAL, time=0, channel=canal)
            )
            track_actual.append(
                mido.Message('note_off', note=nota_rey,
                             velocity=0, time=DURACION_ARPEGIO, channel=canal)
            )
            track_actual.append(
                mido.Message('note_on', note=nota_torre,
                             velocity=VELOCITY_ESPECIAL, time=0, channel=canal)
            )
            track_actual.append(
                mido.Message('note_off', note=nota_torre,
                             velocity=0, time=DURACION_ARPEGIO, channel=canal)
            )

            board.push(move)
            continue

        # --- Caso general ---
        pieza_movida = pieza.piece_type
        nota_base = mapa_piezas_nota.get(pieza_movida, 60)
        velocity  = VELOCITY_NORMAL

        # Duración base según tipo de pieza (articulación)
        if pieza_movida == chess.PAWN:
            duracion = DURACION_NOTA // 2
        elif pieza_movida in (chess.BISHOP, chess.QUEEN):
            duracion = int(DURACION_NOTA * 1.5)
        elif pieza_movida == chess.KING:
            duracion = DURACION_NOTA * 2
        else:
            duracion = DURACION_NOTA

        nota_final = nota_base + octava_offset

        # Captura antes de aplicar el movimiento
        if board.is_capture(move):
            velocity = VELOCITY_CAPTURA

        # Aplicar la jugada
        board.push(move)

        # Jaque después de aplicar la jugada
        if board.is_check():
            velocity = VELOCITY_JAQUE
            duracion = duracion // 2

        # Nota principal
        track_actual.append(
            mido.Message('note_on',  note=nota_final,
                         velocity=velocity, time=0, channel=canal)
        )
        track_actual.append(
            mido.Message('note_off', note=nota_final,
                         velocity=0, time=duracion, channel=canal)
        )

    # 4. Guardar el archivo MIDI
    mid.save(output_filename)

    print("¡Proceso completado!")
    print(f"Archivo MIDI guardado en: {output_filename}")
```

```python
# ----------------------------------------------------------------------
#  Interfaz gráfica (Tkinter): Generar + Play + Stop
# ----------------------------------------------------------------------

# --- INICIALIZAR PYGAME MIXER ---
try:
    # Frecuencia estándar, tamaño bit, canales (estéreo), buffer
    pygame.mixer.init(frequency=44100, size=-16, channels=2, buffer=1024)
    print("Sistema de audio Pygame inicializado.")
except Exception as e:
    print(f"No se pudo inicializar el audio: {e}")

def main():
    root = tk.Tk()
    root.title("Sonificación de Ajedrez - Generador MIDI")
    root.geometry("600x350")  # Ventana un poco más alta para el botón Stop

    # --- Entradas de parámetros ---

    # Tempo
    lbl_tempo = tk.Label(root, text="Tempo (BPM):")
    lbl_tempo.place(x=40, y=40)
    entry_tempo = tk.Entry(root, width=10)
    entry_tempo.insert(0, "50")  # valor por defecto
    entry_tempo.place(x=260, y=40)

    # Instrumento blancas
    lbl_blancas = tk.Label(root, text="Instrumento Blancas (0-127):")
    lbl_blancas.place(x=40, y=80)
    entry_prog_blancas = tk.Entry(root, width=10)
    entry_prog_blancas.insert(0, "48")  # igual a tu código original
    entry_prog_blancas.place(x=260, y=80)

    # Instrumento negras
    lbl_negras = tk.Label(root, text="Instrumento Negras (0-127):")
    lbl_negras.place(x=40, y=120)
    entry_prog_negras = tk.Entry(root, width=10)
    entry_prog_negras.insert(0, "60")
    entry_prog_negras.place(x=260, y=120)
        
    # label con instrucciones
    texto_instrumentos = (
    "Rangos sugeridos para instrumentos MIDI:\n"
    "\n"
    "0 - 10   → Pianos / sonidos suaves\n"
    "11 - 40  → Guitarras / bajos / pads\n"
    "41 - 80  → Cuerdas, brasses, leads\n"
    "81 - 127 → FX y sonidos más extraños"  
    )

    # Imagen debajo del texto de rangos
    # NOTA: Asegúrate de que 'imagen.png' esté en la misma carpeta del notebook
    try:
        if os.path.exists("imagen.png"):
            img_ajedrez = tk.PhotoImage(file="imagen.png")  
            img_ajedrez = img_ajedrez.subsample(10, 10)
            lbl_imagen = tk.Label(root, image=img_ajedrez)
            lbl_imagen.image = img_ajedrez  
            lbl_imagen.place(x=400, y=150)
        else:
            print("Aviso: 'imagen.png' no encontrada, se omite la imagen.")
    except Exception as e:
        print("No se pudo cargar la imagen:", e)


    lbl_info_instr = tk.Label(
        root,
        text=texto_instrumentos,
        justify="left"
    )
    lbl_info_instr.place(x=340, y=40)

    # Label de estado
    lbl_estado = tk.Label(root, text="Sin MIDI generado aún.")
    lbl_estado.place(x=40, y=165)

    # --- Callbacks ---
    def on_generar():
        try:
            tempo = int(entry_tempo.get())
            prog_b = int(entry_prog_blancas.get())
            prog_n = int(entry_prog_negras.get())
        except ValueError:
            messagebox.showerror("Error", "Tempo e instrumentos deben ser números enteros.")
            return

        if not (0 <= prog_b <= 127 and 0 <= prog_n <= 127):
            messagebox.showerror("Error", "Los instrumentos deben estar entre 0 y 127.")
            return

        try:
            generar_midi_desde_pgn(
                pgn_texto=PGN_TEXTO,
                tempo_bpm=tempo,
                prog_blancas=prog_b,
                prog_negras=prog_n,
                output_filename='partida_sonificada_blancas_negras.mid'
            )
            lbl_estado.config(text="MIDI generado. ¡Listo para reproducir!")
        except Exception as e:
            messagebox.showerror("Error al generar MIDI", str(e))

    def on_play():
        nombre_archivo = 'partida_sonificada_blancas_negras.mid'
        if not os.path.exists(nombre_archivo):
            messagebox.showwarning("Sin archivo", "Primero genera el archivo MIDI.")
            return

        try:
            # --- Cargar y Reproducir con Pygame ---
            if not pygame.mixer.get_init():
                pygame.mixer.init()
            
            pygame.mixer.music.load(nombre_archivo)
            pygame.mixer.music.play()
            lbl_estado.config(text="Reproduciendo...")
            
        except Exception as e:
            messagebox.showerror("Error al reproducir", f"Pygame error: {str(e)}")

    def on_stop():
        try:
            if pygame.mixer.music.get_busy():
                pygame.mixer.music.stop()
                lbl_estado.config(text="Reproducción detenida.")
        except Exception as e:
            print(f"Error al detener: {e}")

    # --- Botones ---
    btn_generar = tk.Button(root, text="Generar MIDI", command=on_generar, bg="#dddddd")
    btn_generar.place(x=80, y=210)

    btn_play = tk.Button(root, text="▶ PLAY", command=on_play, bg="#ccffcc", width=10)
    btn_play.place(x=190, y=210)

    btn_stop = tk.Button(root, text="■ STOP", command=on_stop, bg="#ffcccc", width=10)
    btn_stop.place(x=280, y=210)

    root.mainloop()

# Ejecutar la interfaz
if __name__ == "__main__":
    main()
```
